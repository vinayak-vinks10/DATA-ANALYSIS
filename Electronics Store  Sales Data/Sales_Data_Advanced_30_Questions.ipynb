### Q1. Load the CSV file and display first 5 rows.

```python
# Write your answer here
import pandas as pd
data = pd.read_csv("sales_data_project_big.csv")
data.head()
```

### Q2. Display basic information of the dataset.

```python
# Write your answer here
data.info()
```

### Q3. Check for missing values column-wise.

```python
# Write your answer here
data.isna()
```

### Q4. Convert OrderDate column to datetime.

```python
# Write your answer here
data['OrderDate'] = pd.to_datetime(data['OrderDate'], dayfirst=True)
```

### Q5. Create a new column TotalAmount = Quantity * Price.

```python
# Write your answer here
data['TotalAmount']=data['Quantity']*data['Price']
```

### Q6. Find total revenue generated.

```python
# Write your answer here
total_revenue = data['TotalAmount'].sum()
total_revenue
```

### Q7. Find average order value.

```python
# Write your answer here
data['TotalAmount'].mean()
```

### Q8. Find city-wise total revenue.

```python
# Write your answer here
data.groupby("City", as_index=False)["TotalAmount"].sum()

```

### Q9. Find product-wise total revenue.

```python
# Write your answer here
data.groupby("Product", as_index=False)["TotalAmount"].sum()

```

### Q10. Find category-wise total revenue.

```python
# Write your answer here
data.groupby("Category", as_index=False)["TotalAmount"].sum()
```

### Q11. Find top 5 highest revenue orders.

```python
# Write your answer here
data.sort_values("TotalAmount", ascending=False).head(5)

```

### Q12. Find most sold product by quantity.

```python
# Write your answer here
data.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

```

### Q13. Find least sold product by quantity.

```python
# Write your answer here
data.groupby("Product")["Quantity"].sum().sort_values(ascending=True)
```

### Q14. Find month-wise total revenue.

```python
# Write your answer here
data.groupby(data["OrderDate"].dt.to_period("M"))["TotalAmount"].sum()


```

### Q15. Extract month name from OrderDate.

```python
# Write your answer here
data['MonthName'] = data['OrderDate'].dt.month_name()

```

### Q16. Find city and product combination with highest revenue.

```python
combo_revenue_df = (
    data.groupby(["City", "Product"], as_index=False)["TotalAmount"]
    .sum()
    .sort_values("TotalAmount", ascending=False)
)

combo_revenue_df.head(1)
```

### Q17. Find percentage contribution of each city to total revenue.

```python
# Write your answer here
city_rev = data.groupby("City")["TotalAmount"].sum()
(city_rev / city_rev.sum()) * 100

```

### Q18. Find percentage contribution of each product to total revenue.

```python
# Write your answer here
prod_rev = data.groupby("Product")["TotalAmount"].sum()
(prod_rev / prod_rev.sum()) * 100
```

### Q19. Normalize TotalAmount column.

```python
# Write your answer here
data["TotalAmount_Normalized"] = (
    (data["TotalAmount"] - data["TotalAmount"].min()) /
    (data["TotalAmount"].max() - data["TotalAmount"].min())
)

```

### Q20. Find correlation between Quantity and TotalAmount.

```python
# Write your answer here
data["Quantity"].corr(data["TotalAmount"])
```

### Q21. Find orders where TotalAmount is above average.

```python
# Write your answer here
data[data["TotalAmount"] > data["TotalAmount"].mean()]

```

### Q22. Create a pivot table (City vs Category) with total revenue.

```python
# Write your answer here
pd.pivot_table(
    data,
    values="TotalAmount",
    index="City",
    columns="Category",
    aggfunc="sum"
)
```

### Q23. Find rolling 3-order average of TotalAmount.

```python
# Write your answer here
data["Rolling_Avg"] = data["TotalAmount"].rolling(3).mean()
data

```

### Q24. Rank orders based on TotalAmount.

```python
# Write your answer here
data["Revenue_Rank"] = data["TotalAmount"].rank(ascending=False)

```

### Q25. Find cumulative revenue over time.

```python
# Write your answer here
data.sort_values("OrderDate", inplace=True)
data["Cumulative_Revenue"] = data["TotalAmount"].cumsum()

```

### Q26. Detect outlier orders using IQR method.

```python
# Write your answer here
Q1 = data["TotalAmount"].quantile(0.25)
Q3 = data["TotalAmount"].quantile(0.75)
IQR = Q3 - Q1

outliers = data[
    (data["TotalAmount"] < Q1 - 1.5*IQR) |
    (data["TotalAmount"] > Q3 + 1.5*IQR)
]
outliers
```

### Q27. Sort data by OrderDate and TotalAmount.

```python
# Write your answer here
data.sort_values(
    by=["OrderDate", "TotalAmount"],
    ascending=[True, False],
    inplace=True
)
```

### Q28. Filter orders from last 2 months.

```python
# Write your answer here
latest = data["OrderDate"].max()
data[data["OrderDate"] >= latest - pd.DateOffset(months=2)]
```

### Q29. Export cleaned and enriched dataset to CSV.

```python
# Write your answer here
data.to_csv("sales_data_final.csv", index=False)
```
